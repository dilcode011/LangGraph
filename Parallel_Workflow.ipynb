```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class state(TypedDict):
    product:str
    price:str
    reviews:str
```

```python
def price_analysis(state):
    price=state["price"]
    if price<100:
        return "cheap"
    else:
        return "expensive"

def review_analysis(state):
    reviews=state["reviews"]
    if reviews>4:
        return "good"
    else:
        return "bad"

def product_analysis(state):
    product=state["product"]
    return f"product {product} is analyzed"
```

```python
graph=StateGraph(state)
```

```python
graph.add_node("price_analysis",price_analysis)
graph.add_node("review_analysis",review_analysis)
graph.add_node("product_analysis",product_analysis) 
graph.add_edge(START,"price_analysis")
graph.add_edge(START,"review_analysis")
graph.add_edge(START,"product_analysis")
graph.add_edge("price_analysis",END)
graph.add_edge("review_analysis",END)
graph.add_edge("product_analysis",END)
```

```python
workflow=graph.compile()
```

```python
initial_state={"product":"laptop","price":"150000","reviews":"5"}
final_State=workflow.invoke(initial_state)
print(final_State)
```
